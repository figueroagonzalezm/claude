```python
# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-0"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None, temperature=1.0):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "temperature": temperature
    }
    
    if system:
        params["system"] = system
    
    message = client.messages.create(**params)
    return message.content[0].text
```

```python
### EXAMPLE 1 system prompt #####
messages = []
# Without system prompt
#answer = chat(messages)

# With system prompt
system = """
You are a patient math tutor.
Do not directly answer a student's questions.
Guide them to a solution step by step.
"""
add_user_message(messages, "How do I solve 5x + 2 = 3 for x?")
answer = chat(messages, system=system)

answer
```

```python
### EXAMPLE 2 system prompt #####
messages = []

# With system prompt
# if you comment the system prompt, the answer will be more verbose and less concise.
system = """
You are a software developer that writes very concise code.
"""
add_user_message(messages, "Write a Python function that checks a string for duplicate characters.")
answer = chat(messages, system=system)

answer
```

```python
messages = []
add_user_message(messages, "Suggest a plot for a science fiction novel. In two lines.")

# Low temperature - more predictable
#answer = chat(messages, temperature=0.0)

# High temperature - more creative  
answer = chat(messages, temperature=1.0)

answer
```
