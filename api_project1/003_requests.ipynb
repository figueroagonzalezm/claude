```python
# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-0"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)
```

```python
messages=[]

user_text= "Write one sentence description of a fake database"
add_user_message(messages, user_text)

# Stream the response
with client.messages.stream(
    model=model,
    max_tokens=1000,
    messages=messages
) as stream:
    for text in stream.text_stream:
        # Send each chunk to your client as it arrives instead of waiting for the entire response to be generated
        print(text, end="")
```
